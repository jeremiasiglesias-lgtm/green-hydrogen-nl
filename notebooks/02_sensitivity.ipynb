```python
base = {
    "capacity_kw": 10000,          # electrolyser size, kW
    "lifetime": 20,             # years
    "hours_per_year": 8760,       # 365 × 24
    "capex_per_kw": 1950,        # €/kW
    "discount_rate": 0.08,                # discount rate
    "om_fraction": 0.03,         # share of capex per year
    "efficiency": 54.6,            # kWh per kg H2
    "capacity_factor": 0.5,      # share of the year running
    "electricity_price": 87,     # €/MWh
    "reference_capacity_kw": 10_000,   # plant size the capex_per_kw figure applies to
    "scale_exponent": 0.87,             # 1.0 = no economies of scale (placeholder, to source)
    "stack_lifetime_h": 90_000,          # operating hours before the stack must be replaced (alkaline, manufacturer claim)
    "stack_replacement_fraction": 0.48,  # replacement cost as share of capex (arXiv 2509.00136)
    "water_l_per_kg": 10,          # litres demineralised water per kg H2 (9 L stoichiometric + margin)
    "water_price_per_m3": 0.85,    # €/m³, NL demineralised water contract price (Intratec, May 2023)
    "grid_tariff_per_mw_year": 150_000,   # €/MW/yr, TenneT HV transport tariff; 57k with non-firm contract
    "compressor_capex_per_kg_day": 167,   # €/(kg/day) peak capacity, pipeline compressor (Clean Hydrogen JU)
    "compressor_om_fraction": 0.06,       # share of compressor capex per year (arXiv 2509.00136)
    "compression_kwh_per_kg": 0.6,        # kWh/kg, ~30 -> 50 bar into Dutch H2 network (Neuman & Esser: 25->85 bar)
    "compressor_lifetime_years": 14,      # technical lifetime (Clean Hydrogen JU)
}
```

```python
# --- Sourced low/high ranges for each uncertain input (from the assumptions table) ---
# A dictionary whose values are TUPLES: (low, high).
# A tuple is like a list in round brackets, but it can't be changed afterwards.
# That's good here: a range is a fixed pair of numbers.
# --- Sourced low/high ranges for each uncertain input (see assumptions table) ---
ranges = {
    "capex_per_kw":                (1680, 2210),      # €/kW
    "efficiency":                  (50.5, 57.6),      # kWh/kg   (was 47-73)
    "om_fraction":                 (0.02, 0.05),      # share of capex per year
    "discount_rate":               (0.04, 0.10),      # fraction
    "lifetime":                    (20, 25),          # years    (was 10-30)
    "electricity_price":           (77, 96),          # €/MWh    (was 77-87)
    "scale_exponent":              (0.76, 1.0),       # -  kept for the dashboard, skipped in the tornado
    "stack_lifetime_h":            (50_000, 90_000),  # hours    (was 60,000-90,000)
    "water_l_per_kg":              (9, 15),           # L/kg H2
    "water_price_per_m3":          (0.85, 2.19),      # €/m³
    "grid_tariff_per_mw_year":     (116_000, 150_000),# €/MW/yr  (was 57,000-150,000)
    "compressor_capex_per_kg_day": (108, 167),        # €/(kg/day)
    "compression_kwh_per_kg":      (0.3, 2.5),        # kWh/kg
    "compressor_lifetime_years":   (14, 20),          # years
}

# capacity_factor is left out on purpose: it has no sourced range yet (Milestone 3).
```

# 2. Functions

```python
def crf(rate, lifetime):
    growth = (1 + rate) ** lifetime
    return rate * growth / (growth - 1)

def lcoh_breakdown(a):
    lifetime = a["lifetime"]
    capacity_kw = a["capacity_kw"]
    hours_per_year = a["hours_per_year"]
    energy_kwh = capacity_kw * hours_per_year * a["capacity_factor"]
    h2_kg = energy_kwh / a["efficiency"]
    size_ratio = capacity_kw / a["reference_capacity_kw"]
    capex_total = a["reference_capacity_kw"]*a["capex_per_kw"] * size_ratio ** a["scale_exponent"]


    stack_life_years = a["stack_lifetime_h"] / (hours_per_year * a["capacity_factor"])


    replacement_pv = 0
    year = stack_life_years
    while year < lifetime:
        cost = capex_total * a["stack_replacement_fraction"]
        replacement_pv += cost / (1 + a["discount_rate"]) ** year
        year +=  stack_life_years

    stack_per_year = replacement_pv * crf(a["discount_rate"], lifetime)

    capex_per_year = capex_total * crf(a["discount_rate"], lifetime)
    om_per_year = capex_total * a["om_fraction"]
    electricity_per_year = (energy_kwh / 1000) * a["electricity_price"]
    water_per_year = h2_kg * (a["water_l_per_kg"] / 1000) * a["water_price_per_m3"]   # L -> m³
    grid_per_year = (capacity_kw/1000) * a["grid_tariff_per_mw_year"]        # kW -> MW

    peak_kg_per_day = capacity_kw / a["efficiency"] * 24                    # kg/day at full power
    compressor_capex = peak_kg_per_day * a["compressor_capex_per_kg_day"]
    compressor_replacement_pv = 0
    year = a["compressor_lifetime_years"]
    while year < lifetime:
        compressor_replacement_pv += compressor_capex / (1 + a["discount_rate"]) ** year
        year += a["compressor_lifetime_years"]

    compression_energy_cost = h2_kg * a["compression_kwh_per_kg"] / 1000 * a["electricity_price"]   # kWh -> MWh
    compression_per_year = ((compressor_capex + compressor_replacement_pv) * crf(a["discount_rate"], lifetime)
                            + compressor_capex * a["compressor_om_fraction"]
                            + compression_energy_cost)

    return {
        "capex": capex_per_year/h2_kg,
        "om": om_per_year/h2_kg,
        "electricity": electricity_per_year/h2_kg,
        "stack": stack_per_year / h2_kg,
        "water": water_per_year / h2_kg,
        "grid": grid_per_year / h2_kg,
        "compression": compression_per_year / h2_kg,
        "total": (capex_per_year + om_per_year + electricity_per_year+ stack_per_year + water_per_year + grid_per_year + compression_per_year)/h2_kg,
    }

def lcoh_from(a):
    """LCOH in €/kg from the dictionary of assumptions."""
    return lcoh_breakdown(a)["total"]
```

# 3. Electricity price chart

```python
import matplotlib.pyplot as plt

prices = list(range(0, 101, 10))
lcoh_cf05 = []
lcoh_cf09 = []

for price in prices:
    s = base.copy()
    s["electricity_price"] = price
    s["capacity_factor"] = 0.5
    lcoh_cf05.append(lcoh_from(s))

    t = base.copy()
    t["electricity_price"] = price
    t["capacity_factor"] = 0.9
    lcoh_cf09.append(lcoh_from(t))

plt.plot(prices, lcoh_cf05, label="Capacity factor 0.5")
plt.plot(prices, lcoh_cf09, label="Capacity factor 0.9")
plt.axhline(2, linestyle="--", color="grey", label="Grey H2 (placeholder)")
plt.xlabel("Electricity price (€/MWh)")
plt.ylabel("LCOH (€/kg)")
plt.title("Green hydrogen cost vs electricity price")
plt.grid(True)
plt.legend()
plt.show()
```

# 4. Tornado Chart

```python
rows = []

to_test = ["efficiency", "capacity_factor","capex_per_kw","electricity_price","discount_rate","lifetime","om_fraction"]
for name in to_test:

    low = base.copy()
    low[name] = base[name] * 0.8
    high = base.copy()
    high[name] = base[name] * 1.2

    lcoh_low = lcoh_from(low)
    lcoh_high = lcoh_from(high)
    swing = abs(lcoh_high - lcoh_low)

    rows.append((swing, name, lcoh_low, lcoh_high))

rows.sort()

for swing, name, lcoh_low, lcoh_high in rows:
    start = min(lcoh_low, lcoh_high)
    plt.barh(name, swing, left=start)

plt.axvline(lcoh_from(base), color="black", label=f"Base case (€{lcoh_from(base):.2f}/kg)")
plt.xlabel("LCOH (€/kg)")
plt.title("Sensitivity of LCOH to ±20% changes")
plt.xlim(6, 10)
plt.legend()
plt.show()
```

**Chart:** shows how much the LCOH changes when any input changes by ±20%, one at a time. Efficiency moves the LCOH the most: more kWh per kg means more electricity to produce one kg of hydrogen, and fewer kg from the same plant.

**Excluded and Kept Uncertainties:**
- `hours_per_year`: hours per year and capacity factor scale identically, since the capacity factor is a fraction of the yearly hours. Hours per year is also a fixed fact, so there is no uncertainty.
- `capacity_kw`: plant size is a design choice, not an uncertainty. In this model it has no effect, because all costs and the hydrogen output scale equally with increasing size they cancel. The model is therefore missing economies of scale, so I added a scaling rule (exponent currently 1.0, a placeholder until sourced).
- `lifetime` is kept: it is both a design choice and an uncertainty. Some factors outside your control may affect lifetime, such as legislation, but you can build more expensive electrolysers that last longer.

**Caveat:** some of these factors are more volatile than others, so comparing them on an equal percentage basis is biased. Some conditions could easily change by 20% or more, while others are unlikely to change even by smaller percentages. The ranking will be redone with sourced, realistic ranges.

#5 Plant size and economies of scale


```python
test = base.copy()
test["capacity_kw"] = 100_000
test["scale_exponent"] = 0.8
print(f"LCOH: €{lcoh_from(test):.2f}")
```

**Result:** a 100 MW plant with scale exponent 0.8 gives an LCOH of about €8.88/kg, compared with €10.07/kg for the 100 MW base case.

**Explanation:** with the scaling factor, as our plant scales by a factor of 10, the capex only scales by 10^0.8 ≈ 6.3. This means capex has scaled slower than H2 production, causing a lower LCOH: capex per kg falls to about 63% of the base case. O&M per kg falls too, because it is a percentage of capex. Electricity per kg stays the same (€2.75), since every kg still needs 55 kWh. So economies of scale only reduce the fixed costs.

**Note:** the exponent 0.8 is a test value, not yet sourced. Electrolysers are modular, so the real exponent may be closer to 1.

# 7. Tornado chart with sourced ranges

```python
SKIP = ["scale_exponent"]   # no effect at 10 MW (size_ratio = 1), tested separately in section 5
MIN_SWING = 0.05            # €/kg: smaller swings are printed in the table but not drawn

base_lcoh = lcoh_from(base)     # €/kg, the centre line of the chart
rows = []

for name, (low_value, high_value) in ranges.items():
    if name in SKIP:            # 'in' checks whether a value appears in a list
        continue                # 'continue' jumps straight to the next round of the loop

    low_case = base.copy()
    low_case[name] = low_value
    lcoh_low = lcoh_from(low_case)      # €/kg with this input at its LOW value

    high_case = base.copy()
    high_case[name] = high_value
    lcoh_high = lcoh_from(high_case)    # €/kg with this input at its HIGH value

    swing = abs(lcoh_high - lcoh_low)   # €/kg, width of the bar
    rows.append((swing, name, lcoh_low, lcoh_high))

rows.sort()    # smallest swing first, so the biggest ends up on top of the chart

# Full table, including the small ones we won't draw
for swing, name, lcoh_low, lcoh_high in reversed(rows):
    print(f"{name:28} low input -> €{lcoh_low:5.2f}  high input -> €{lcoh_high:5.2f}  swing €{swing:.2f}/kg")

# Keep only the inputs that matter for the chart.
# This is a 'list comprehension': a one-line loop that builds a new list
# from the rows whose swing is at least MIN_SWING.
big_rows = [row for row in rows if row[0] >= MIN_SWING]

for i, (swing, name, lcoh_low, lcoh_high) in enumerate(big_rows):
    if i == 0:
        low_label = "Input at low value"
        high_label = "Input at high value"
    else:
        low_label = None
        high_label = None
    plt.barh(name, lcoh_low - base_lcoh,  left=base_lcoh, color="tab:blue",   label=low_label)
    plt.barh(name, lcoh_high - base_lcoh, left=base_lcoh, color="tab:orange", label=high_label)

plt.axvline(base_lcoh, color="black", label=f"Base case (€{base_lcoh:.2f}/kg)")
plt.xlabel("LCOH (€/kg)")
plt.title("Sensitivity of LCOH to sourced input ranges")
plt.legend()
plt.tight_layout()
plt.show()
```

### 7. Sensitivity with sourced ranges: results and interpretation

**What the chart shows.** Each bar shows how far the LCOH moves when one input goes from its sourced low to its sourced high value, with all other inputs at their base values. Blue = input at its low value, orange = input at its high value. Base case: €10.07/kg.

**Ranking (swing in €/kg):** efficiency 1.28, discount rate 1.10, electricity price 1.05, capex 0.87, O&M 0.73, stack lifetime 0.49, grid tariff 0.42, compression energy 0.19, compressor capex 0.06.

**Left out of the chart:**
- Swings below €0.05/kg: compressor lifetime (0.03), plant lifetime (0.02), water use and water price (0.01 each). They are printed in the table but don't change any conclusion.
- Scale exponent: no effect here, because the plant (10 MW) is the same size as the reference the capex is quoted for. size_ratio = 10,000 / 10,000 = 1, and 1 to any power is 1. Economies of scale are tested separately in section 5.
- Capacity factor: no sourced range yet. Milestone 2 showed it is one of the strongest drivers, so this is the biggest gap in the chart. Milestone 3 (hourly prices) will set it.

**How to read the colours.** For most inputs a higher value raises the cost (orange on the right). Stack lifetime is reversed: a 50,000 h stack has to be replaced once, around year 11 at CF 0.5, adding about €0.49/kg. A 90,000 h stack lasts the full 20 years.

**Scenarios.**
- 2022 crisis power price: LCOH rises to €18.57/kg, almost double the base case.
- Non-firm grid contract: looks like it saves €1.16/kg (→ €8.91), but with ~15% curtailment the plant makes fewer kg to spread its fixed costs over, so the real saving is €0.44/kg (→ €9.63). I assumed curtailment cuts running hours by 15%.

**Key takeaways.**
1. With realistic ranges no single input dominates: efficiency, financing cost and electricity price each move the LCOH by about €1/kg.
2. Financing matters as much as technology. Lowering the discount rate from 8% to 4% (for example through subsidies or government guarantees) saves about €0.71/kg, about the same as the best realistic efficiency.
3. The biggest electricity risk is not the normal year-to-year range but rare crisis years. That is why developers lock in long-term power contracts (PPAs), and why Milestone 3 looks at which hours to run.
4. Cheaper grid access only helps if the lost running hours are counted: the non-firm contract keeps only ~40% of its headline saving.
5. Even at the best value of any single input, the LCOH stays above €9.3/kg, far above the placeholder €2/kg grey benchmark. No single lever closes the gap; it needs several at once (Milestones 3, 4 and 6).

**Limitations.** One-at-a-time analysis ignores inputs that move together: for example, high gas prices raise Dutch power prices *and* the cost of grey hydrogen at the same time. The electricity price is still a flat yearly average and the capacity factor is fixed at 0.5. The efficiency high value comes mostly from smaller units, so it may be pessimistic for a 10 MW plant.
