### Certified-green shortcut: running only in low-price hours (≤ €26.64/MWh)

**Why this scenario.** Under EU rules for renewable hydrogen (Delegated Regulation 2023/1184), electricity used in any hour with a day-ahead price at or below €20/MWh, or below 0.36 × the EU ETS price, automatically meets the time-matching condition. With the 2025 average ETS price of €74/t the limit is max(20, 0.36 × 74) = €26.64/MWh. Running only in those hours is the simplest way to approach certified green hydrogen.

**Result (NL, 2025):**
- Hours at or below €26.64/MWh: 1,211 h, so capacity factor 0.14
- Average power price in those hours: −€1.6/MWh (the plant is paid to use power)
- LCOH: €18.93/kg, vs €7.48/kg for cost-optimal grid operation (CF 0.9)
- "Cost of being green" by this route: about €11.45/kg, or 2.5× the grid optimum
- First version used the €20 limit only (1,144 h, €19.96/kg); corrected after rereading the regulation (C5)

**Why it is so expensive despite near-free power.** The plant is idle 86% of the year. Capex, O&M and the grid tariff are fixed yearly costs, so they are spread over only about a seventh of the hydrogen the plant could make. Cheap electricity cannot make up for low utilisation.

**Limitations.**
- The low-price rule only covers time matching. A real project still needs the PPA framework (additionality, geographical correlation), so this is a bound, not a full business case.
- Uses the yearly average ETS price; hourly ETS prices would be more exact.
- One year of data only. As solar and wind grow, the number of cheap hours is likely to rise, which would lower this cost.
- Assumes the plant can switch on and off freely every hour (no minimum load or restart costs).

**Takeaway.** Green hydrogen cannot be built on cheap hours alone: it needs certified renewable power for most of the year. That is what the PPA scenarios (Milestone 3b) test.

# Milestone 3: summary

**Question.** With real Dutch hourly prices, how many hours should a grid-connected electrolyser run, and what does that do to the LCOH?

**Method.** ENTSO-E day-ahead prices for the Netherlands, 2025. From 1 October 2025 the market uses 15-minute prices, so these were averaged to hourly values (8,760 hours, mean €86.8/MWh, matching published figures). The plant runs only in the cheapest X% of hours; for each share the model calculates the average price paid and the LCOH.

**Price facts 2025.** 584 negative-price hours, median €90.4/MWh, 1,144 hours at or below €20/MWh and 1,211 at or below the EU low-price limit (€26.64/MWh). The duration curve has a small block of very cheap hours (solar/wind), a long flat middle (€60–130/MWh) and a few hundred spike hours.

**Results.**
- Cost-optimal operation: capacity factor 0.9 (7,884 h/yr, €77.4/MWh) gives €7.48/kg, against €10.07/kg in the old base case (CF 0.5, flat €87/MWh).
- About €2.37 of the €2.59/kg saving comes from running more hours; only €0.22/kg from skipping the most expensive hours. Utilisation matters far more than picking cheap hours, because fixed costs (capex, O&M and especially the per-MW grid tariff) are paid whether the plant runs or not.
- The optimum is flat: anywhere between CF 0.7 and 1.0 costs within €0.22/kg of the best case.
- With cheaper electrolysers (€1,000/kW) the optimum shifts down to CF 0.8 (≈€6.40/kg): when idle hours cost less, skipping expensive hours pays off more.
- Certified-green shortcut (only hours ≤ €26.64/MWh, the EU low-price rule): CF 0.14, €18.93/kg, about 2.5× the grid optimum.

**Business takeaway.** A grid-connected Dutch electrolyser should run most of the year. But cost-optimal grid operation is not certified green hydrogen, and running only in cheap hours is far too expensive. Green hydrogen needs certified renewable power for most hours, which is what Milestone 3b tests with renewable PPAs.

**Limitations.** One year of prices (2025); perfect knowledge of prices (realistic for day-ahead); the plant can switch on and off every hour with no minimum load or restart costs; Dutch energy taxes and levies not yet included.

```python
# =====================================================================
# Milestone 3b: renewable production profiles (NL, 2025)
# =====================================================================
# Actual electricity generation per technology (MW), every time step of 2025.
# This is a DataFrame: a table with one COLUMN per technology.
gen_file = DATA + "nl_generation_2025.csv"
if os.path.exists(gen_file):                   # saved before -> load it (no API call)
    # header=[0, 1]: the file has TWO header rows
    # (technology name, then "Actual Aggregated" / "Actual Consumption")
    gen_raw = pd.read_csv(gen_file, header=[0, 1], index_col=0)
    gen_raw.index = pd.to_datetime(gen_raw.index, utc=True).tz_convert("Europe/Amsterdam")
else:                                          # first time only -> download and save
    gen_raw = client.query_generation("NL", start=start, end=end)
    gen_raw = gen_raw[gen_raw.index < end]     # drop the 2026 value
    gen_raw.to_csv(gen_file)

# Before using any dataset, LOOK at it first
print(gen_raw.shape)          # expect (35039, 20)
print(gen_raw.columns)        # which technologies are there?
print(gen_raw.head())
```

```python
# --- 1. Keep only the columns we need -------------------------------
gen = gen_raw.copy()

# ENTSO-E sometimes gives two column levels, e.g. ("Solar", "Actual Aggregated").
# isinstance(...) checks which format we got; .xs(...) keeps only the "Actual Aggregated" level.
if isinstance(gen.columns, pd.MultiIndex):
    gen = gen.xs("Actual Aggregated", axis=1, level=1)

gen = gen[["Wind Offshore", "Wind Onshore", "Solar"]]   # a list of names inside [] selects several columns
gen = gen.resample("h").mean()                          # MW, one value per hour (same as for prices)

# --- 2. Installed capacity (MW) of each technology in 2025 ----------
cap_raw = client.query_installed_generation_capacity("NL", start=start, end=end)
print(cap_raw)                                          # look at it first!
capacity = cap_raw.iloc[0][["Wind Offshore", "Wind Onshore", "Solar"]]   # first row = 2025, MW

# --- 3. Per-MW profile: production ÷ installed capacity -------------
# Pandas divides each column by the matching capacity automatically.
profile = gen / capacity                                # MW produced per MW installed (0 to ~1)

print(gen.mean())
print(profile.head())
print("Capacity factor per technology:")
print(profile.mean().round(3))                          # average over the year = capacity factor
```

### Data check: ENTSO-E coverage of Dutch solar and onshore wind

**Capacity factors 2025 from ENTSO-E:** offshore wind 0.377, onshore wind 0.113, solar 0.002.
Typical values are roughly 0.40–0.50 (offshore), 0.25–0.30 (onshore) and ~0.10 (solar).

**Diagnosis.** Offshore wind is plausible. Solar and onshore are far too low. Average reported solar production is only 57 MW against 29,282 MW installed. Most Dutch solar and part of onshore wind is connected to regional grids, and ENTSO-E's generation data appears to capture only part of it, while the capacity figure counts everything. Production ÷ capacity therefore gives a CF that is far too low.

**Decision.** Milestone 3b version 1 uses offshore wind only: the data is complete (all offshore farms connect to TenneT's high-voltage grid), and it matches the typical Dutch hydrogen PPA. Solar and onshore profiles to be added later from a weather-based source (e.g. Renewables.ninja).
